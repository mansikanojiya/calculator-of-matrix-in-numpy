```python
import numpy as np

def add_matrices(A, B):
    rows, cols = A.shape
    result = np.zeros((rows, cols))

    for i in range(rows):
        for j in range(cols):
            result[i][j] = A[i][j] + B[i][j]

    return result
```

```python
def subtract_matrices(A, B):
    rows, cols = A.shape
    result = np.zeros((rows, cols))

    for i in range(rows):
        for j in range(cols):
            result[i][j] = A[i][j] - B[i][j]

    return result
```

```python
def transpose_matrix(A):
    rows, cols = A.shape
    result = np.zeros((cols, rows))

    for i in range(rows):
        for j in range(cols):
            result[j][i] = A[i][j]

    return result
```

```python
def multiply_matrices(A, B):
    rows_A, cols_A = A.shape
    rows_B, cols_B = B.shape

    if cols_A != rows_B:
        raise ValueError("Matrix multiplication not possible")

    result = np.zeros((rows_A, cols_B))

    for i in range(rows_A):
        for j in range(cols_B):
            for k in range(cols_A):
                result[i][j] += A[i][k] * B[k][j]

    return result
```

```python
def show_menu():
    print("\nMatrix Calculator")
    print("1. Add matrices")
    print("2. Subtract matrices")
    print("3. Transpose matrix")
    print("4. Multiply matrices")
    print("5. Exit")
```

```python
def input_matrix(rows, cols):
    matrix = np.zeros((rows, cols))
    print(f"Enter elements for a {rows}x{cols} matrix:")

    for i in range(rows):
        for j in range(cols):
            matrix[i][j] = float(input(f"Element [{i}][{j}]: "))

    return matrix
```

```python
while True:
    show_menu()
    choice = input("Choose an option (1-5): ")

    if choice == "1":
        r = int(input("Number of rows: "))
        c = int(input("Number of columns: "))
        A = input_matrix(r, c)
        B = input_matrix(r, c)
        print("Result:")
        print(add_matrices(A, B))

    elif choice == "2":
        r = int(input("Number of rows: "))
        c = int(input("Number of columns: "))
        A = input_matrix(r, c)
        B = input_matrix(r, c)
        print("Result:")
        print(subtract_matrices(A, B))

    elif choice == "3":
        r = int(input("Number of rows: "))
        c = int(input("Number of columns: "))
        A = input_matrix(r, c)
        print("Result:")
        print(transpose_matrix(A))

    elif choice == "4":
        r1 = int(input("Rows of Matrix A: "))
        c1 = int(input("Columns of Matrix A: "))
        r2 = int(input("Rows of Matrix B: "))
        c2 = int(input("Columns of Matrix B: "))

        if c1 != r2:
            print("Matrix multiplication not possible")
            continue

        A = input_matrix(r1, c1)
        B = input_matrix(r2, c2)
        print("Result:")
        print(multiply_matrices(A, B))

    elif choice == "5":
        print("Goodbye!")
        break

    else:
        print("Invalid choice. Try again.")
```
